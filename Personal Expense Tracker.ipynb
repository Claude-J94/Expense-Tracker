###  1. Setup

```python
# Import necessary libraries
import csv
from datetime import datetime
```

### 2. File Initialization

```python
# Define file name
FILE_NAME = 'expenses.csv'

# Ensure the CSV file exists with headers
def init_file():
    try:
        with open(FILE_NAME, 'x', newline='') as file:
            writer = csv.writer(file)
            writer.writerow(['Date', 'Category', 'Amount', 'Description'])
            print("File initialized.")
    except FileExistsError:
        print("File already exists.")
        
init_file()
```

### ➕ 3. Add an Expense

```python
def add_expense(date, category, amount, description):
    with open(FILE_NAME, 'a', newline='') as file:
        writer = csv.writer(file)
        writer.writerow([date, category, amount, description])
    print("Expense added.")
    
# Example:
# add_expense('2025-05-02', 'Food', 12.5, 'Lunch at cafe')
```

📖 4. View All Expenses

```python
def view_expenses():
    with open(FILE_NAME, 'r') as file:
        reader = csv.reader(file)
        for row in reader:
            print(row)

# view_expenses()
```

### 📊 5. Get Summary Report

```python
def summary_report():
    total = 0
    categories = {}
    
    with open(FILE_NAME, 'r') as file:
        reader = csv.DictReader(file)
        for row in reader:
            amount = float(row['Amount'])
            total += amount
            category = row['Category']
            categories[category] = categories.get(category, 0) + amount
            
    print(f"Total Expenses: ${total:.2f}")
    print("Expenses by Category:")
    for cat, amt in categories.items():
        print(f"  {cat}: ${amt:.2f}")
        
# summary_report()
```

### ❌ 6. Delete Expense

```python
def delete_expense(index):
    expenses = []

    # Load all expenses
    with open(FILE_NAME, 'r') as file:
        reader = csv.reader(file)
        expenses = list(reader)

    # Show current list with indices
    print("Current Expenses:")
    for i, row in enumerate(expenses):
        print(f"{i}: {row}")
    
    # Check for valid index
    if index <= 0 or index >= len(expenses):
        print("Invalid index. Cannot delete header or non-existent row.")
        return

    # Delete the selected row
    deleted = expenses.pop(index)
    print(f"Deleted: {deleted}")

    # Write the updated list back to the file
    with open(FILE_NAME, 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerows(expenses)

# Example usage:
# delete_expense(2)
```

### ✏️ Modify Expense Function

```python
def modify_expense(index):
    expenses = []

    # Load all expenses
    with open(FILE_NAME, 'r') as file:
        reader = csv.reader(file)
        expenses = list(reader)

    # Show current list
    print("Current Expenses:")
    for i, row in enumerate(expenses):
        print(f"{i}: {row}")

    # Check for valid index
    if index <= 0 or index >= len(expenses):
        print("Invalid index. Cannot modify header or non-existent row.")
        return

    # Show current values
    current = expenses[index]
    print(f"\nCurrent values: Date: {current[0]}, Category: {current[1]}, Amount: {current[2]}, Description: {current[3]}")

    # Ask for new values (press Enter to keep existing)
    new_date = input(f"New date (YYYY-MM-DD) [Leave blank to keep '{current[0]}']: ") or current[0]
    new_category = input(f"New category [Leave blank to keep '{current[1]}']: ") or current[1]
    new_amount = input(f"New amount [Leave blank to keep '{current[2]}']: ") or current[2]
    new_description = input(f"New description [Leave blank to keep '{current[3]}']: ") or current[3]

    # Validate amount and date
    try:
        float(new_amount)
        datetime.strptime(new_date, '%Y-%m-%d')
    except ValueError:
        print("Invalid input. Update failed.")
        return

    # Update and save
    expenses[index] = [new_date, new_category, new_amount, new_description]

    with open(FILE_NAME, 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerows(expenses)

    print("Expense updated successfully.")
```

### 📋 7. Interactive Menu (Basic Console UI)

```python
def show_menu():
    print("1. Add Expense")
    print("2. View Expenses")
    print("3. Summary Report")
    print("4. Delete Expense")
    print("5. Modify Expense")
    print("6. Exit")


def main():
    init_file()
    
    while True:
        show_menu()
        choice = input("Enter your choice (1-6): ")

        if choice == '1':
            date = input("Enter date (YYYY-MM-DD): ")
            category = input("Enter category: ")
            amount = input("Enter amount: ")
            description = input("Enter description: ")
            try:
                amount = float(amount)
                datetime.strptime(date, '%Y-%m-%d')  # Validate date format
                add_expense(date, category, amount, description)
            except ValueError:
                print("Invalid input. Please try again.")

        elif choice == '2':
            view_expenses()

        elif choice == '3':
            summary_report()

        elif choice == '4':
            try:
                index = int(input("Enter the row number to delete: "))
                delete_expense(index)
            except ValueError:
                print("Invalid index. Try again.")
        
        elif choice == '5':
            try:
                index = int(input("Enter the row number to modify: "))
                modify_expense(index)
            except ValueError:
                print("Invalid index. Try again.")


        elif choice == '6':
            print("Goodbye!")
            break

        else:
            print("Invalid choice. Please enter 1-6.")

# Run the main menu loop
main()
```
